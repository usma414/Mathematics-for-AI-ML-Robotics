##Dot Product Of Vectors

```python
a = np.array([1,2,3,4,5])
b = np.array([6,7,8,9,10])
c = np.array([11,12,13,14,15])
```

```python
np.dot(a, b)
```

###Basic Properties of Dot Product

- It is **Commutative**:   a.b = b.a
- It is **Distributive**:  a.(b+c) = a.b + a.c

```python
# Commutative Proof
np.dot(b, a)
```

```python
# Distributive Proof

leftside = np.dot(a, b+c)
print(leftside)    #Answer = 335
rightside = np.dot(a, b) + np.dot(a,c)
print(rightside)

if(leftside == rightside):
    print(True)
```

```python
import numpy as np
from numpy import linalg as lng
```

```python
a = np.array([2,3]);
lng.norm(a)    #Gives Magnitude of a vector
```

##Scalar and Vector Projection

- A vector projection of a **vector a** onto **vector b** is the **orthogonal projection a onto b**.

- Formula for projection = proj = **(a.b / |b|^2) * b**

```python
a = np.array([10,20,30,40,50])
b = np.array([60,70,80,90,100])

vector_projection = (np.dot(a, b) / (lng.norm(b)* lng.norm(b))) * b
print(vector_projection)
```

##Basis, Linear Independence, Span

- e1 and e2 are unit vectors along x-axes and y axes
- 4e1 - 4e2 = a = [4,-3] = 4i - 3j
- Unit Vectors or basis vector form a basis for space
- Any vector can be written as a linear combination of these two vectors

**Basic Properties of Basis Vectors:**

- are ***linearly Independent*** of each other
- ***Span*** the whole space
- Aren't ***Unique***
